## <span style="color:Orange">Overview</span>

Understanding how and why we are here is one of the fundamental questions for the ages. Part of the answer to this question lies in the origins of galaxies, such as our own Milky Way. Yet questions remain about how the Milky Way (or any of the other ~100 billion galaxies in our Universe) was formed and has evolved. Galaxies come in all shapes, sizes and colors: from beautiful spirals to huge ellipticals. “Kevin Schawinski, previously an astrophysicist at Oxford University and co-founder of Galaxy Zoo, described the problem that led to Galaxy Zoo's creation when he was set the task of classifying the morphology of more than 900,000 galaxies by eye that had been imaged by the Sloan Digital Sky Survey at the Apache Point Observatory in New Mexico, USA.”

### <span style="color:LightGreen">Question 01</span>

What is the Sky Survey? What motivated scientists to create such a survey?

The Sloan Digital Sky Survey is a pioneering ground-based astronomical survey operated from the Apache Point Observatory in New Mexico that began in 2000 and is known for its novel instrumentation and data reduction/storage techniques. Its dedicated 2.5m wide-field optical telescope fitted with a multi-filter (*u,g,r,i,z*) scanning CCD camera enables it to efficiently capture large areas of sky, and its groundbreaking multi-fiber spectrograph enables several hundred source spectra to be extracted at once.

The survey has imaged over 35% of the sky and made photometric observations of ~1B celestial objects and spectra of ~4 million. Many scientific motivations necessitating such a large, detailed map of the night sky spurred the creation of the survey. These motivations include:

*   **Universe-Scale Estimation:** Whereas prior astronomical data covered smaller or shallower sections of sky often chosen beforehand for their inferred properties of interest, the SDSS covered much larger swaths of the sky without prior selection. This enabled scientists to collect enough data covering a large enough span to make statistical inferences about universe-scale properties, including the distribution, general properties, and evolution of various celestial bodies and structures.
*   **Large- & Small-Scale Structure:** A 3D map of 1/3rd of the sky would enable scientists much greater insight into the underlying structure of the universe across many scales. Things like the universe's dark matter distribution, galaxy clustering, and the substructure of the Milky Way were able to be further understood from the SDSS's data.

### <span style="color:LightGreen">Question 01</span>

What is the Sky Survey? What motivated scientists to create such a survey?

>The Sloan Digital Sky Survey is a pioneering ground-based astronomical survey operated from the Apache Point Observatory in New Mexico that began in 2000 and is known for its novel instrumentation and data reduction/storage techniques. Its dedicated 2.5m wide-field optical telescope fitted with a multi-filter (*u,g,r,i,z*) scanning CCD camera enables it to efficiently capture large areas of sky, and its groundbreaking multi-fiber spectrograph enables several hundred source spectra to be extracted at once.

>The survey has imaged over 35% of the sky and made photometric observations of ~1B celestial objects and spectra of ~4 million. Many scientific motivations necessitating such a large, detailed map of the night sky spurred the creation of the survey. These motivations include:

>*   **Universe-Scale Estimation:** Whereas prior astronomical data covered smaller or shallower sections of sky often chosen beforehand for their inferred properties of interest, the SDSS covered much larger swaths of the sky without prior selection. This enabled scientists to collect enough data covering a large enough span to make statistical inferences about universe-scale properties, including the distribution, general properties, and evolution of various celestial bodies and structures.
>*   **Large- & Small-Scale Structure:** A 3D map of 1/3rd of the sky would enable scientists much greater insight into the underlying structure of the universe across many scales. Things like the universe's dark matter distribution, galaxy clustering, and the substructure of the Milky Way were able to be further understood from the SDSS's data.
>* **Spectrography & Multi-wavelength Photometry:** The wide range of filters through which images can be taken enables the estimation of photometric redshifts (photo-z), and the extraction of spectra helps better understand the fine properties of celestial objects, including early-universe dark matter distributions.


### <span style="color:LightGreen">Question 02</span>

Galaxy Zoo guides its citizen scientists through a nested decision tree - this is what constitutes the classification process. ​Why is it important to be able to classify galaxies? Why are the different shapes of galaxies so important? What information is provided from different galaxy colors?

>Galaxy Zoo is a project concerned with identifying the morphology, or observed shape, of galaxies as seen in survey photo data. Classifying galaxies (by their morphology) is crucial for astronomy as it informs us about many relevant qualities of the galaxy. Morphology can reflect a galaxy's evolution history–its formation, the stage of its life, potential previous mergers, star activity, and beyond. Investigating the structure of galaxies from their morphology can uncover the star formation rate, gas content, and other relevant factors on the internal dynamics of the galaxy, as well as insight into the galaxy's dark matter halo and its role in the galaxy's shape. Alternatively to looking at the morphologies themselves, observing the proportion of different kinds of morphologies (elliptical, spiral, irregular, etc) allows us to estimate population-level statistics on galaxies.

>Different galaxy morphologies generally correspond to different physical properties of the galaxies, often related to their formation and evolution. The different shapes include, but are far from limited to:
>*   **Elliptical:** Smooth, featureless spherical/elongated galaxies that may appear as bright extended spots in images. They tend to be populated by older, red stars with minimal new star formation. Research has estimated such galaxies are often the result of violent past mergers long ago, with much of the internal gas and dust of the galaxy having already been consumed by the existing stars.
>*   **Spiral:** These galaxies are almost always a rotating disk of stars, dust, and gas with many distinguishable "arms" around a dense spherical bulge of stars surrounding a central supermassive black hole. They are much more active in terms of star formation and contain substantial gas for continued stellar production.
>*   **Irregular:** For galaxies with more complex morphologies and irregular shape, we often infer a merger is taking place between two galaxies. They lack defined arms and other features, though are rich in dust, gas, and star production due to the collisions between celestial bodies within the mergers.

Galaxy colors (derived from the relative magnitudes of images across filters and their signature on the spectra) inform us primarily of the level of activity within the galaxy and, as a result, its approximate age.
>*   Bluer galaxies tend to indicate the presence of hot, young stars and generally active ongoing star formation. Thus, we typically expect to see this signature from spiral galaxies and within irregular galaxies.
>*   Redder galaxies tend to indicate older, cooler stars and the absence of gas and new star formation. We expect these in elliptical galaxies which have used up all their star-birthing fuel already and are cooling off.
>*   Irregular galaxies can exhibit many colors(including both red and blue) as a result of ongoing galaxy mergers since the two galaxies may be at entirely different stages of evolution and host different kinds of stars. However, we certainly expect prominent signatures in the bluer ranges due to the significant new star formation caused by the galaxy collision.
>*   Even within non-irregular galaxies, we can observe variations in color. For example, in spiral galaxies the dense gas and dust found in the arms lends itself well to active star formation, resulting in a bluer color on the outside compared to towards the center.

### <span style="color:LightGreen">Question 03</span>

After reading a thorough reading of the paper in [[1]](https://arxiv.org/pdf/1308.3496.pdf), ​describe what constitutes the decision tree. What are the conditions described in the decision tree?

>As a part of Galaxy Zoo, ~40-50 participants were asked to identify individual galaxies according to a series of sequential questions as a decision tree:
1. Is the object a smooth galaxy, a galaxy with features/disk or a star?
    - Smooth → Q7
    - Features or Disk → Q2
    - Star or Artifact → End

2. Could this be a disk viewed edge-on?
    - Yes → Q9
    - No → Q3

3. Is there a sign of a bar feature through the centre of the galaxy?
    - Yes → Q4
    - No → Q4

4. Is there any sign of a spiral arm pattern?
    - Yes → Q10
    - No → Q5

5. How prominent is the central bulge, compared with the rest of the galaxy?
    - No bulge → Q6
    - Just noticeable → Q6
    - Obvious → Q6
    - Dominant → Q6

6. Is there anything odd?
    - Yes → Q8
    - No → End

7. How rounded is it?
    - Completely Rounded → Q6
    - In-Between → Q6
    - Cigar-Shaped → Q6
    
8. Is the odd feature a ring, or is the galaxy disturbed or irregular?
    - Ring → End
    - Lens or Arc → End
    - Disturbed → End
    - Irregular → End
    - Other → End
    - Merger → End
    - Dust Lane → End

9. Does the galaxy have a bulge at its centre? If so, what shape?
    - Rounded → Q6
    - Boxy → Q6
    - No Bulge → Q6

10. How tightly wound do the spiral arms appear?
    - Tight → Q11
    - Medium → Q11
    - Loose → Q11

11. How many spiral arms are there?
    - 1 → Q5
    - 2 → Q5
    - 3 → Q5
    - 4 → Q5
    - \>4 → Q5
    - Can't Tell → Q5

Volunteers answer only the questions for each galaxy that are relevant to that specific galaxy, i.e. along the "flowchart" path their answers take them. So, each proportion of votes following the first question is a conditional probability. That is, the proportion of responses to a question are multiplied by the proportion of the response of the preceding question, and so on until returning to the first. The output of the decision tree then gives not a single label or output, but a distribution over the possible outcomes, qualitatively capturing the uncertainty and disagreement in responses for each source.
